# Hybrid Search for Beginners

## What is Hybrid Search?

Combines **two search methods** to find the best results:

1. **Keyword Search (BM25)** - Finds exact words

2. **Semantic Search (Vectors)** - Understands meaning

Think of it like asking a librarian (keyword) AND a book expert (semantic) to find you the perfect book!

---

## Understanding Documents in RAG: page_content vs metadata

When building a RAG (Retrieval Augmented Generation) system, documents have two parts:

### 1. `page_content` (The Main Content)

- **What it is:** The actual text that answers user questions

- **How it's used:** This gets embedded and searched to find relevant context

- **RAG Example:** "To reset your password, click on 'Forgot Password' and enter your email..."

### 2. `metadata` (Extra Information)

- **What it is:** Tracking info ABOUT the content (source, timestamp, category, etc.)

- **How it's used:** For filtering, citation, and tracking - NOT for searching

- **RAG Example:** `{"source": "user_manual.pdf", "page": 15, "section": "Account Settings"}`

### How Does Search Work in RAG?

```
User Question: "How do I reset my password?"
         │
         ▼
┌────────────────────────────────────────┐
│  RAG Retrieval Step                   │
│  Search ONLY looks at page_content    │
│  Metadata is NOT searched!             │
└────────────────────────────────────────┘
         │
         ▼
🔍 Keyword Search: Finds "reset" + "password" in page_content
🧠 Semantic Search: Finds "recover account access" (similar meaning)
         │
         ▼
    Top K Documents Retrieved
    (with metadata for citation)
         │
         ▼
    Sent to LLM as Context
         │
         ▼
    LLM Generates Answer
```

**Important for RAG:** 

- ✅ Retrieval searches `page_content` only

- ❌ Retrieval does NOT search `metadata`

- 📋 Metadata helps cite sources in the final answer

- 🤖 Both content AND metadata can be sent to the LLM

```
**RAG Example:**
Document(
    page_content="To reset password, go to Settings > Security > Reset",  # ← Searched & sent to LLM
    metadata={"source": "manual.pdf", "page": 15}  # ← Not searched, but useful for citation
)
```

When user asks "reset password", RAG:

1. Finds this document by searching `page_content`

2. Sends the content to LLM as context

3. LLM generates answer + can cite "Source: manual.pdf, page 15" using metadata

## Simple RAG Code Example

```
# Build RAG with Hybrid Search
def rag_with_hybrid_search(user_question, knowledge_base):
    # Step 1: Hybrid retrieval - find relevant context
    keyword_results = bm25_retriever.get_relevant_documents(user_question)
    semantic_results = vector_retriever.get_relevant_documents(user_question)

    # Step 2: Combine results
    hybrid_results = ensemble_retriever.get_relevant_documents(user_question)

    # Step 3: Build context from page_content
    context = "\n\n".join([doc.page_content for doc in hybrid_results])

    # Step 4: Build citations from metadata
    sources = [doc.metadata.get("source") for doc in hybrid_results]

    # Step 5: Send to LLM
    prompt = f"Context: {context}\n\nQuestion: {user_question}\n\nAnswer:"
    answer = llm.generate(prompt)

    return {
        "answer": answer,
        "sources": sources  # Citation info from metadata
    }

# Example usage
response = rag_with_hybrid_search(
    "How do I reset my password?", 
    knowledge_base
)
print(response["answer"])
print(f"Sources: {response['sources']}")
```

```
filename = "/Users/vinotganesan/Learning/LLM & AGENTS/STANFORD/SLIDES/Chapter 3-LLM.pdf"
print(f"Loading PDF file: {filename}")

def read_pdf_text(pdf_path):
    all_text = ""
    # Use a context manager to ensure the PDF file is closed properly
    with pdfplumber.open(pdf_path) as pdf:
        # Loop through all the pages in the document
        for page in pdf.pages:
            # Extract text from the current page
            text = page.extract_text()
            if text:
                all_text += text + "\n"
    return all_text

total_text = read_pdf_text(filename)
# Split it up
text_splitter = RecursiveCharacterTextSplitter(chunk_size=400, chunk_overlap=50)
chunks = text_splitter.split_text(total_text)  # Use split_text() for strings, not split_documents()
chunks = text_splitter.create_documents(chunks)
```

```
from langchain.embeddings.base import Embeddings
from pydantic import BaseModel
from typing import Any, List, Mapping, Optional
import requests
import numpy as np  # Added missing import


class MLServerEmbedding(Embeddings, BaseModel):
    max_batch_size: int = 32
    url: str

    """ public """

    def embed_documents(self, texts: List[str]) -> List[List[float]]:
        text_num = len(texts)
        result = []
        for i in range(0, text_num, self.max_batch_size):
            result += self._get_embedding(texts[i: i + self.max_batch_size])
        return result

    def embed_query(self, text: str) -> List[float]:
        return self._get_embedding([text])[0]

    """ private """

    @staticmethod
    def _wrap_payload(text_list):
        return {
            "inputs": [
                {
                    "name": "input",
                    "shape": [len(text_list)],
                    "datatype": "str",
                    "data": text_list
                }
            ]
        }

    @staticmethod
    def _parse_response(response):
        if response.status_code != 200:
            raise Exception(response)
        outputs = response.json()["outputs"][0]
        return np.array(outputs["data"]).reshape(outputs["shape"]).tolist()

    def _get_embedding(self, text_list: List[str]) -> List[List[float]]:
        return self._parse_response(requests.post(url=self.url,
                                                  json=self._wrap_payload(text_list),
                                                  headers={"Content-Type": "application/json"},
                                                  params={}))
```

```
embedding_model = "all-mpnet-base--fccb5"
embedding_url = ""+ embedding_model +"/v2/models/"+ embedding_model + "/infer"
embeddings = MLServerEmbedding(url=embedding_url)
```

```
# Test if embedding_model (MLServerEmbedding) is working
test_text = "This is a test sentence for MLServer embeddings."
try:
    test_embedding_mlserver = embeddings.embed_query(test_text)
    print(f"✅ MLServerEmbedding is working!")
    print(f"Embedding dimension: {len(test_embedding_mlserver)}")
    print(f"First 5 values: {test_embedding_mlserver[:5]}")
except Exception as e:
    print(f"❌ MLServerEmbedding failed: {e}")
```

# Maximal Marginal Relevance (MMR)

Maximal Marginal Relevance (MMR) is a method used in information retrieval to select documents that are both relevant to the query and diverse with respect to the previously selected documents. This approach helps in reducing redundancy and increasing the coverage of different aspects of the query in the selected documents.

```
# Vector retriever — fetch top 10 candidates for reranking
retriever_mmr = vectordb.as_retriever(search_type="mmr", search_kwargs={"k": 10})

print("✅ Vector (MMR) Retriever: k=10")
```

```
# Test the vanilla retriever with a sample query
query = "What are the main types of Transformer models?"

# Get search results
results = retriever_mmr.get_relevant_documents(query)

print(f"Query: {query}")
print(f"\nFound {len(results)} relevant documents:\n")
print("=" * 80)

for i, doc in enumerate(results):
    print(f"\n📄 Result {i}:")
    print(f"Content: {doc.page_content[:200]}...")  # Show first 200 chars
    print(f"Metadata: {doc.metadata}")
    print("-" * 80)
```

# BM25 - Keyword search

BM25, or Okapi BM25, is a ranking function used in **information retrieval to estimate the relevance of documents to a given search query.** It is an enhancement of the traditional TF-IDF method, which considers the frequency of terms within a document (term frequency) and the rarity of terms across all documents (inverse document frequency) to compute relevance scores.

```
# BM25 retriever — fetch top 10 candidates for reranking
retriever_BM25 = BM25Retriever.from_documents(chunks, k=10)

print("✅ BM25 Retriever: k=10")
```

```
# Define the human message prompt template
human_message_template = PromptTemplate.from_template(
    """
    You are an experienced assistant specializing in question-answering tasks. 
    Utilize the provided context to respond to the question. 
    If the answer is unknown, always state 'I don't know.' 
    Never provide an answer you are unsure about and ensure it is concise.
    Your answer must be comprehensive and contain all of the relevant details in the Context.
    \nQuestion: {question} \nContext: {context} \nAnswer:
    """
)

human_message_prompt_template = HumanMessagePromptTemplate(prompt=human_message_template)

chat_prompt_template = ChatPromptTemplate(
    input_variables=['context', 'question'],
    messages=[human_message_prompt_template]
)
```

```
rag_chain = chat_prompt_template | llm | StrOutputParser()
```

```
ensemble_retriever = EnsembleRetriever(
    retrievers=[retriever_mmr, retriever_BM25], weights=[0.6, 0.4]
)
```

# Using Flash ReRanker

```
from langchain.retrievers.contextual_compression import ContextualCompressionRetriever
from langchain_community.document_compressors import FlashrankRerank
from langchain.chains import RetrievalQA
from flashrank import Ranker

ranker = Ranker(model_name="ms-marco-MiniLM-L-12-v2")

# top_n=3 → reranker keeps only the 3 best chunks from the 20 candidates
# (10 from vectorDB + 10 from BM25, deduplicated by EnsembleRetriever)
compressor = FlashrankRerank(top_n=3)

compression_retriever = ContextualCompressionRetriever(
    base_compressor=compressor,
    base_retriever=ensemble_retriever,
)

chain = RetrievalQA.from_chain_type(llm=llm, retriever=compression_retriever)

# Quick sanity-check: show what the reranker actually kept
reranked_docs = compression_retriever.invoke(question)
print(f"Candidates fetched  : 10 (vector) + 10 (BM25) = up to 20 after dedup")
print(f"After reranking     : {len(reranked_docs)} chunks kept\n")
for i, doc in enumerate(reranked_docs):
    print(f"── Chunk {i+1} ──")
    print(doc.page_content[:200])
    print()

result = chain.invoke(question)
print("Answer:", result["result"])
```
