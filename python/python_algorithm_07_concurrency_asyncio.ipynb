# Concurrency — asyncio Patterns

Adobe Interview Prep | 14 YOE

asyncio is single-threaded cooperative multitasking:

  - One event loop, one OS thread, many coroutines.

  - A coroutine voluntarily yields control at every `await`.

  - Zero race conditions on shared Python objects (no GIL fights).

  - Best for I/O-bound work (HTTP, DB, file, websockets).
