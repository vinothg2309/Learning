## 1. What is Mem0?

**Mem0** is an intelligent, self-improving memory layer for AI assistants and agents.

### Key Purpose:

- Provides **persistent memory** across conversations

- Enables **personalized AI** that remembers user preferences

- Solves LLMs' **fixed context window limitation**

- Reduces **costs and latency** while improving accuracy

```
### Think of it as:
Regular LLM:  "Who are you?" → Forgets everything after session ends
Mem0-powered: "Who are you?" → Remembers you, your preferences, past conversations
```

```
### Architecture:
┌─────────────────────────────────────┐
│         Your AI Application         │
└────────────┬────────────────────────┘
             │
             ▼
┌─────────────────────────────────────┐
│          Mem0 Memory Layer          │
│  ┌──────────┐  ┌─────────────────┐ │
│  │  Vector  │  │  Graph Memory   │ │
│  │  Search  │  │  (Relations)    │ │
│  └──────────┘  └─────────────────┘ │
└─────────────────────────────────────┘
             │
             ▼
┌─────────────────────────────────────┐
│     Storage (Vector DB + Graph)     │
└─────────────────────────────────────┘
```

## 2. Why Use Mem0?

```
### Problem Without Mem0:
# Session 1
User: "I'm allergic to peanuts"
AI: "Got it!"

# Session 2 (Next day)
User: "Recommend a snack"
AI: "How about peanut butter cookies?"  # ❌ Forgot!
```

```
### Solution With Mem0:
# Session 1
User: "I'm allergic to peanuts"
AI: "Got it!" → Mem0 stores: "User allergic to peanuts"

# Session 2 (Next day)
User: "Recommend a snack"
Mem0 retrieves: "User allergic to peanuts"
AI: "How about apple slices or carrot sticks?"  # ✅ Remembers!
```

### Key Benefits:

| Benefit | Impact |

|---------|--------|

| **+26% Accuracy** | Better responses using past context |

| **91% Faster** | Reduced latency vs full-context methods |

| **90% Lower Cost** | Less tokens needed for context |

| **Multi-Session** | Remembers across days, weeks, months |

| **Multi-User** | Separate memories per user |

| **Self-Improving** | Automatically consolidates and updates |

## 6. Core Concepts

### Memory Hierarchy:

```
┌────────────────────────────────────────────────┐
│         Organizational Memory                  │  Weeks-Forever
│  "Company policy: 30-day return window"        │
└────────────────────────────────────────────────┘
             ▲
             │
┌────────────────────────────────────────────────┐
│            User Memory                         │  Weeks-Forever
│  "Alice prefers vegetarian food"               │
└────────────────────────────────────────────────┘
             ▲
             │
┌────────────────────────────────────────────────┐
│          Session Memory                        │  Minutes-Hours
│  "Meeting scheduled at 3 PM today"             │
└────────────────────────────────────────────────┘
             ▲
             │
┌────────────────────────────────────────────────┐
│       Conversation Memory                      │  Real-time
│  "Current message: What's for lunch?"          │
└────────────────────────────────────────────────┘
```

### Memory Types:

```
**1. Factual Memory:**
# User facts and preferences
"User is allergic to peanuts"
"User lives in San Francisco"
"User's account tier: Premium"
```

```
**2. Episodic Memory:**
# Summaries of interactions
"On Dec 20, user asked about travel to Japan"
"Previous issue #123 resolved on Dec 15"
```

```
**3. Semantic Memory:**
# Concept relationships (Graph)
"User knows Alex" → "Alex likes coffee" → "User's friend prefers espresso"
```

### Core Operations:

```
# 1. ADD - Store new memories
memory.add(messages, user_id="alice")

# 2. SEARCH - Retrieve relevant memories
results = memory.search(query, filters={"user_id": "alice"})

# 3. UPDATE - Modify existing memory
memory.update(memory_id="123", data="Updated preference")

# 4. DELETE - Remove memories
memory.delete(memory_id="123")
memory.delete_all(filters={"user_id": "alice"})
```

### Memory Lifecycle:

```
User Input → Extract → Consolidate → Store → Retrieve → Use
                ↓                      ↓
           LLM Analysis        Vector + Graph DB
```

## 8. Basic Usage Examples

### Example 1: Simple Add & Search

```
from mem0 import Memory
import os

# Configure with proper model (reuse env vars from cell 13)
config = {
    "llm": {
        "provider": "openai",
        "config": {
            "model": "gpt-5-nano",
            "temperature": 0.1,
            "max_tokens": 2000
        }
    },
    "embedder": {
        "provider": "openai",
        "config": {
            "model": "text-embedding-3-small"
        }
    }
}

# Initialize with config
m = Memory.from_config(config)

# Add memories from conversation
messages = [
    {"role": "user", "content": "Hi, I'm Alex. I love basketball and gaming."},
    {"role": "assistant", "content": "Hey Alex! Nice to meet you. I'll remember your interests."}
]

result = m.add(messages, user_id="alex")
print("Added memories:", result)

# Search memories
results = m.search(
    query="What are Alex's hobbies?",
    filters={"user_id": "alex"},
    limit=3
)

print("\nSearch Results:")
for mem in results:
    print(f"- {mem['memory']} (score: {mem.get('score', 0):.2f})")
```

### Example 2: Complete Chat Application

```
from mem0 import Memory
from openai import OpenAI
import os

# Configure Mem0 with proper model
mem0_config = {
    "llm": {
        "provider": "openai",
        "config": {
            "model": "gpt-5-nano",
            "temperature": 0.1,
            "max_tokens": 2000
        }
    },
    "embedder": {
        "provider": "openai",
        "config": {
            "model": "text-embedding-3-small"
        }
    }
}

# Initialize
memory = Memory.from_config(mem0_config)
client = OpenAI()

def chat_with_memory(user_id: str, message: str) -> str:
    """
    Chat function with persistent memory
    
    Args:
        user_id: Unique user identifier
        message: User's input message
    
    Returns:
        AI assistant's response
    """
    # 1. Retrieve relevant memories
    relevant_memories = memory.search(
        query=message,
        filters={"user_id": user_id},
        limit=3
    )
    
    # 2. Format context from memories
    context = "\n".join([mem['memory'] for mem in relevant_memories]) if relevant_memories else "No prior context"
    
    # 3. Generate response with context
    response = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[
            {
                "role": "system",
                "content": f"You are a helpful assistant. Use this context about the user:\n{context}"
            },
            {"role": "user", "content": message}
        ]
    )
    
    assistant_response = response.choices[0].message.content
    
    # 4. Store the interaction
    memory.add([
        {"role": "user", "content": message},
        {"role": "assistant", "content": assistant_response}
    ], user_id=user_id)
    
    return assistant_response

# Test the chat function
print("Session 1:")
response1 = chat_with_memory("alice", "I'm allergic to peanuts and I love pasta.")
print(f"AI: {response1}\n")

print("Session 2 (simulating next day):")
response2 = chat_with_memory("alice", "What should I have for dinner?")
print(f"AI: {response2}")
```

## 9. LangGraph Integration

### Pattern Overview:

```
┌─────────────────────────────────────────────┐
│         LangGraph State Machine             │
│                                             │
│  START → chatbot_node → chatbot_node → ... │
│              │              │               │
│              ▼              ▼               │
│          Mem0 Retrieval & Storage           │
└─────────────────────────────────────────────┘
```

### Complete LangGraph + Mem0 Example:

```
import os
from typing import Annotated, TypedDict, List
from langchain_core.messages import HumanMessage, AIMessage, BaseMessage
from langchain_openai import ChatOpenAI
from langgraph.graph import StateGraph, START, END
from langgraph.graph.message import add_messages
from mem0 import Memory

# Corporate proxy endpoint configuration
api_key = os.environ.get('UNIFIED_LLM_KEY', 'your-unified-llm-key')
base_url = ""

# Set environment variables for Mem0 to use corporate proxy
os.environ["OPENAI_API_KEY"] = api_key
os.environ["OPENAI_BASE_URL"] = base_url


# Initialize LLM with corporate proxy
llm = ChatOpenAI(
    model="gpt-4o-mini",
    temperature=0,
    max_tokens=256,
    api_key=api_key,
    base_url=base_url
)

# Initialize Mem0 - uses OPENAI_BASE_URL env var
mem0_config = {
    "llm": {
        "provider": "openai",
        "config": {
            "model": "gpt-5-nano",  # Changed to gpt-5-nano (available model)
            "temperature": 0.1,
            "max_tokens": 2000
            # Uses OPENAI_API_KEY and OPENAI_BASE_URL from environment
        }
    },
    "embedder": {
        "provider": "openai",
        "config": {
            "model": "text-embedding-3-small"
            # Uses OPENAI_API_KEY and OPENAI_BASE_URL from environment
        }
    }
}

memory = Memory.from_config(mem0_config)

# Define State
class State(TypedDict):
    messages: Annotated[List[BaseMessage], add_messages]
    user_id: str

def chatbot_node(state: State) -> State:
    """
    Main chatbot node with Mem0 integration
    
    Flow:
    1. Retrieve relevant memories
    2. Generate response with context
    3. Store interaction in Mem0
    """
    user_id = state["user_id"]
    current_message = state["messages"][-1].content
    
    # Step 1: Retrieve memories
    relevant_memories = memory.search(
        query=current_message,
        filters={"user_id": user_id},
        limit=5
    )
    
    # Step 2: Format context
    context = "\n".join([f"- {mem['memory']}" for mem in relevant_memories]) if relevant_memories else "No prior context"
    
    # Step 3: Generate response
    system_message = f"""You are a helpful assistant with access to user's memory.
    
Relevant context about the user:
{context}

Use this context to provide personalized responses."""
    
    messages = [
        {"role": "system", "content": system_message},
        {"role": "user", "content": current_message}
    ]
    
    response = llm.invoke(messages)
    
    # Step 4: Store interaction
    memory.add([
        {"role": "user", "content": current_message},
        {"role": "assistant", "content": response.content}
    ], user_id=user_id)
    
    # Return updated state
    return {
        "messages": [AIMessage(content=response.content)],
        "user_id": user_id
    }

# Build graph
workflow = StateGraph(State)
workflow.add_node("chatbot", chatbot_node)
workflow.add_edge(START, "chatbot")
workflow.add_edge("chatbot", END)

graph = workflow.compile()

print("✅ LangGraph + Mem0 agent ready with corporate proxy!")
print(f"   Using OPENAI_BASE_URL: {os.environ.get('OPENAI_BASE_URL')}")
print(f"   Using Mem0 model: gpt-5-nano")
```

### Test the Agent:

```
# Conversation 1
print("=== Conversation 1 ===")
result1 = graph.invoke({
    "messages": [HumanMessage(content="Hi, I'm Sarah and I love hiking in the mountains.")],
    "user_id": "sarah"
})
print(f"AI: {result1['messages'][-1].content}\n")

# Conversation 2 (same session)
print("=== Conversation 2 ===")
result2 = graph.invoke({
    "messages": [HumanMessage(content="I also prefer gluten-free meals.")],
    "user_id": "sarah"
})
print(f"AI: {result2['messages'][-1].content}\n")

# Conversation 3 (next day - simulated)
print("=== Conversation 3 (Next Day) ===")
result3 = graph.invoke({
    "messages": [HumanMessage(content="Can you suggest weekend activities and lunch options?")],
    "user_id": "sarah"
})
print(f"AI: {result3['messages'][-1].content}")
```

### Advanced: Multi-Agent with Shared Memory

```
from langgraph.graph import StateGraph, START, END
from mem0 import Memory

# Configure with proper model
mem0_config = {
    "llm": {
        "provider": "openai",
        "config": {
            "model": "gpt-5-nano",
            "temperature": 0.1,
            "max_tokens": 2000
        }
    },
    "embedder": {
        "provider": "openai",
        "config": {
            "model": "text-embedding-3-small"
        }
    }
}

memory = Memory.from_config(mem0_config)

class MultiAgentState(TypedDict):
    messages: Annotated[List[BaseMessage], add_messages]
    user_id: str
    agent_id: str  # Different agents can share or separate memories

def support_agent(state: MultiAgentState) -> MultiAgentState:
    """Support agent with access to user's support history"""
    user_id = state["user_id"]
    message = state["messages"][-1].content
    
    # Retrieve support-related memories
    memories = memory.search(
        query=message,
        filters={"user_id": user_id},
        limit=5
    )
    
    # Filter for support category if metadata exists
    support_mems = [m for m in memories if m.get('metadata', {}).get('category') == 'support']
    context = "\n".join([m['memory'] for m in support_mems]) if support_mems else "No past support issues"
    
    # Generate support response
    response = llm.invoke([
        {"role": "system", "content": f"You are a support agent. Past issues:\n{context}"},
        {"role": "user", "content": message}
    ])
    
    # Store with support category
    memory.add(
        [{"role": "user", "content": message}, {"role": "assistant", "content": response.content}],
        user_id=user_id,
        metadata={"category": "support", "agent": "support_agent"}
    )
    
    return {"messages": [AIMessage(content=response.content)]}

def sales_agent(state: MultiAgentState) -> MultiAgentState:
    """Sales agent with access to user's purchase history"""
    user_id = state["user_id"]
    message = state["messages"][-1].content
    
    # Retrieve sales-related memories
    memories = memory.search(
        query=message,
        filters={"user_id": user_id},
        limit=5
    )
    
    # Filter for sales category if metadata exists
    sales_mems = [m for m in memories if m.get('metadata', {}).get('category') == 'sales']
    context = "\n".join([m['memory'] for m in sales_mems]) if sales_mems else "No purchase history"
    
    # Generate sales response
    response = llm.invoke([
        {"role": "system", "content": f"You are a sales agent. Purchase history:\n{context}"},
        {"role": "user", "content": message}
    ])
    
    # Store with sales category
    memory.add(
        [{"role": "user", "content": message}, {"role": "assistant", "content": response.content}],
        user_id=user_id,
        metadata={"category": "sales", "agent": "sales_agent"}
    )
    
    return {"messages": [AIMessage(content=response.content)]}

print("✅ Multi-agent system with categorized memories ready!")
```

### 2. Error Handling

```
from mem0 import Memory
import logging

logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

# Configure with proper model
config = {
    "llm": {
        "provider": "openai",
        "config": {
            "model": "gpt-5-nano",
            "temperature": 0.1,
            "max_tokens": 2000
        }
    },
    "embedder": {
        "provider": "openai",
        "config": {
            "model": "text-embedding-3-small"
        }
    }
}

memory = Memory.from_config(config)

def safe_add_memory(content: str, user_id: str):
    """Add memory with error handling"""
    try:
        result = memory.add(content, user_id=user_id)
        logger.info(f"Memory added for user {user_id}")
        return result
    except Exception as e:
        logger.error(f"Failed to add memory: {e}")
        return None

def safe_search_memory(query: str, user_id: str, limit: int = 5):
    """Search memory with error handling and fallback"""
    try:
        results = memory.search(
            query=query,
            filters={"user_id": user_id},
            limit=limit
        )
        return results
    except Exception as e:
        logger.error(f"Memory search failed: {e}")
        return []  # Return empty list as fallback

# Test
safe_add_memory("Test memory", "user123")
results = safe_search_memory("test", "user123")
print(f"Found {len(results)} memories")
```
