# Concurrency — Multiprocessing Patterns

Adobe Interview Prep | 14 YOE

Multiprocessing spawns separate OS processes — each gets its own Python

interpreter and GIL.  True CPU parallelism on multi-core machines.

Use when:

  • CPU-bound: image/video encoding, ML inference, compression, crypto

  • Need to bypass GIL for pure-Python number crunching

  • Isolating unstable third-party code (crashes don't kill main process)

Covers:

  1.  Process basics — Process, start, join

  2.  ProcessPoolExecutor (preferred over Pool for modern Python)

  3.  Shared memory — Value, Array (mutable across processes)

  4.  Manager — shared dict/list (complex objects across processes)

  5.  Pipe — bidirectional IPC between two processes

  6.  Queue (multiprocessing) — N-producer / M-consumer

  7.  Pool.map vs Pool.imap vs Pool.starmap

  8.  Chunk strategy — optimal chunk_size for large workloads

  9.  Inter-process signalling with Event and Semaphore

  10. Mixing asyncio + ProcessPoolExecutor
