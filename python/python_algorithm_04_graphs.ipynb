# Graphs — Senior Interview Patterns

Adobe Interview Prep | 14 YOE

Covers:

  1.  Graph representations (adjacency list, matrix)

  2.  DFS — iterative & recursive, cycle detection

  3.  BFS — shortest path in unweighted graph

  4.  Topological Sort — Kahn's (BFS) & DFS post-order

  5.  Union-Find (Disjoint Set Union) with path compression + rank

  6.  Dijkstra's Algorithm — shortest path in weighted graph

  7.  Bellman-Ford — negative weights, negative cycle detection

  8.  Island problems (number of islands, max area, surrounded regions)

  9.  Clone Graph

  10. Course Schedule (detect cycle in directed graph)
