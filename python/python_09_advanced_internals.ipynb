## Descriptor Protocol Deep Dive

```
class Descriptor:
    def __set_name__(self, owner, name):
        self.name = name

    def __get__(self, obj, objtype=None):
        if obj is None:
            return self
        return getattr(obj, f'_{self.name}', None)

    def __set__(self, obj, value):
        setattr(obj, f'_{self.name}', value)
```

## Method Resolution Order (MRO)

```
class A: pass
class B(A): pass
class C(A): pass
class D(B, C): pass

print(D.__mro__)
# D → B → C → A → object
```

## Key Interview Insights

1. **Variable Lookup** - LEGB scope (Local, Enclosing, Global, Builtin) is determined at compile time for optimization.

2. **Attribute Access** - Instance `__dict__` is checked before the class `__dict__`, but data descriptors take priority.

3. **Method Binding** - Methods are actually functions stored on the class. When accessed through an instance, the descriptor protocol binds them.

4. **Integer Interning** - Small integers (-5 to 256) are cached for performance. String interning happens at compile time for literals.

5. **Import Caching** - Modules are cached in `sys.modules`. Re-importing doesn't re-execute the module.

6. **Generator State** - Generators maintain their own frame object, which stores local variables and execution position.
